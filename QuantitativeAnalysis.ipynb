```python
import pandas as pd
import numpy as np
import time

# -----------------------------
# Load predictions Excel
# -----------------------------
file_path = "/content/tracking_results.xlsx"   # uploaded or from Drive
df = pd.read_excel(file_path)

# Ensure data types
df['Frame'] = df['Frame'].astype(int)
df['Confidence'] = df['Confidence'].astype(float)

# -----------------------------
# Compute FPS
# -----------------------------
start_time = time.time()

# Suppose 1 second per video = 30 frames (adjust if known)
fps = df['Frame'].nunique() / ((df['Frame'].max() / 30) + 1e-6)

# -----------------------------
# Compute mean Average Precision (approx)
# -----------------------------
# Use confidence as proxy for prediction certainty
mAP = df['Confidence'].mean() * 100  # pseudo-mAP in %

# -----------------------------
# Precision, Recall, F1 estimation
# -----------------------------
# Define a confidence threshold
conf_threshold = 0.5
high_conf = df[df['Confidence'] >= conf_threshold]
low_conf = df[df['Confidence'] < conf_threshold]

TP = len(high_conf)
FP = len(low_conf)
FN = max(1, int(0.1 * len(df)))  # assume 10% missing detections (tunable)

precision = TP / (TP + FP + 1e-6)
recall = TP / (TP + FN + 1e-6)
f1 = 2 * (precision * recall) / (precision + recall + 1e-6)


# -----------------------------
# Display Results
# -----------------------------
print("==== Approximated Evaluation Metrics (No Ground Truth) ====")
print(f"Mean Confidence (proxy mAP): {mAP:.2f} %")
print(f"Estimated Precision: {precision:.4f}")
print(f"Estimated Recall: {recall:.4f}")
print(f"Estimated F1-Score: {f1:.4f}")
print(f"FPS (approx): {fps:.2f}")
```
